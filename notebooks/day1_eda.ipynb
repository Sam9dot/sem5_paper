```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/system_logs_dataset.csv")
```

```python
print("SHAPE")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
print("CLASS BALANCE (is_attack)")
print(df["is_attack"].value_counts())
print(df["is_attack"].value_counts(normalize=True) * 100)
```

```python
print("ATTACK TYPE BREAKDOWN")
print(df["attack_name"].value_counts())
```

```python
print("NULLS PER COLUMN")
nulls = df.isnull().sum()
print(nulls[nulls > 0])
```

```python
print("NULLS PER COLUMN")
nulls = df.isnull().sum()
print(nulls[nulls > 0])
```

```python
print("DUPLICATE ROWS")
print(f"Duplicate rows: {df.duplicated().sum()}")
```

```python
print("COLUMN DTYPES")
print(df.dtypes)
```

```python
print("'disk' UNIQUE VALUES (categorical feature)")
print(df["disk"].unique())
```

```python
DROP_COLS = ["timestamp", "attack_name"]
print(f"\nColumns flagged to drop before modeling: {DROP_COLS}")
```

```python
summary_lines = [
    "# Data Summary — Day 1\n",
    f"- Rows: {df.shape[0]}, Columns: {df.shape[1]}\n",
    f"- Class balance (is_attack): {dict(df['is_attack'].value_counts())}\n",
    f"- Class balance %: {dict((df['is_attack'].value_counts(normalize=True) * 100).round(2))}\n",
    f"- Duplicate rows: {df.duplicated().sum()}\n",
    f"- Columns with nulls: {list(nulls[nulls > 0].index)}\n",
    f"- Categorical column 'disk' values: {list(df['disk'].unique())}\n",
    f"- Columns to drop before modeling: {DROP_COLS}\n",
    "- Null handling decision: attack_name nulls are expected (blank for normal rows) — drop column entirely, not an imputation case.\n",
]
 
with open("../results/data_summary.md", "w") as f:
    f.writelines(summary_lines)
 
print("\nSaved results/data_summary.md")
```
